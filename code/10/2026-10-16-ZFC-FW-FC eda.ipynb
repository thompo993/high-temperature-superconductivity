```python
import numpy as np
import matplotlib.pyplot as plt

pb_temps_sq = pb_temps**2


slope, y_intercept = np.polyfit(pb_fields, pb_temps_sq, 1)

# extended regime for pb_feilds plotting 
pb_fields_extended = np.linspace(0, pb_fields[-1] + 150, 50)
fitted_line = slope * pb_fields_extended + y_intercept

x_intercept = - y_intercept/slope

plt.plot(pb_fields, pb_temps_sq, linestyle="", marker="o", label="Data")
plt.plot(pb_fields_extended, fitted_line, linestyle="-", label=f"Y-Intercept:{y_intercept:.2f}\nGradient:{slope:.2f}\nX-intercept:{x_intercept} ")

plt.title(r"261016 - Pb $T^2$ vs $H$")
plt.xlabel("Magnetic Field (Oe)")
plt.ylabel(r"Temperature$^2$ (K$^2$)")
plt.legend()
plt.ylim(0,55)
plt.xlim(0,800)
plt.show()
```
