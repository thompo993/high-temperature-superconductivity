```python
# Reload the FW dataset because `bi` currently contains the FC dataset
fw = read_squid_file(
    r"C:\Users\olive\OneDrive - University of Bristol\Final Project\Data\Ollie_and_Ben\2026\10\FeSe_bo001_MvsT_FW_VariousFields\2026-10-22_FeSe_bo001_MvsT_FW_Various_Field_00001.dat"
)
fc_sorted = bi.sort_values("Temperature (K)")
fig, ax = plt.subplots(figsize=(6, 4))

fw_sorted = fw.sort_values("Temperature (K)")

ax.plot(
    fw_sorted["Temperature (K)"],
    fw_sorted["DC Moment Fixed Ctr (emu)"],
    color="tab:blue",
    label="FW"
)

ax.plot(
    bi["Temperature (K)"],
    bi["DC Moment Fixed Ctr (emu)"],
    color="tab:red",
    s=12,
    label="FC"
)

ax.set_title("Pb M vs T at 5 Oe")
ax.set_xlabel("Temperature (K)")
ax.set_ylabel("DC Moment Fixed Ctr (emu)")
ax.legend()
plt.show()
```

```python
import ipywidgets as widgets
from IPython.display import display

x_col = "Temperature (K)"
y_col = "DC Moment Fixed Ctr (emu)"

fw_toggle = widgets.Checkbox(value=True, description="FW")
fc_toggle = widgets.Checkbox(value=True, description="FC")

fig, ax = plt.subplots(figsize=(6, 4))

def update_plot(*args):
    ax.clear()

    if fw_toggle.value:
        fw_sorted = fw.sort_values(x_col)
        ax.plot(
            fw_sorted[x_col],
            fw_sorted[y_col],
            color="tab:blue",
            label="FW"
        )

    if fc_toggle.value:
        fc_sorted = bi.sort_values(x_col)
        ax.plot(
            fc_sorted[x_col],
            fc_sorted[y_col],
            color="tab:red",
            label="FC"
        )

    ax.set_title("Pb M vs T at 5 Oe")
    ax.set_xlabel("Temperature (K)")
    ax.set_ylabel("DC Moment Fixed Ctr (emu)")
    ax.legend()
    fig.canvas.draw_idle()

fw_toggle.observe(update_plot, names="value")
fc_toggle.observe(update_plot, names="value")

display(widgets.HBox([fw_toggle, fc_toggle]))
update_plot()
```

```python
fig_static, ax_static = plt.subplots(figsize=(6,4))

fw_sorted = fw.sort_values(x_col)
fc_sorted = bi.sort_values(x_col)

ax_static.plot(
    fw_sorted[x_col],
    fw_sorted[y_col],
    color="tab:blue",
    label="Field Warming (FW)"
)

ax_static.plot(
    fc_sorted[x_col],
    fc_sorted[y_col],
    color="tab:red",
    label="Field Cooling (FC)"
)

ax_static.set_title(
    "FeSe (sample: bo001)\nMagnetic Moment vs Temperature, FW & FC at 10 Oe"
)
ax_static.set_xlabel(x_col)
ax_static.set_ylabel(y_col)
ax_static.legend(loc = 'best')

plt.show()
```
